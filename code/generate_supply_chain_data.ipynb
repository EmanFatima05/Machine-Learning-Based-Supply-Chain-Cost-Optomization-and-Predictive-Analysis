# Supply Chain Data Generation
This notebook generates all dimension and fact tables for the supply chain database.

```python
import pandas as pd
import numpy as np
import random
import os
from datetime import datetime, timedelta

random.seed(42)
np.random.seed(42)

os.makedirs('data', exist_ok=True)
```

```python
# ── dim_date ────
start_date = datetime(2019, 1, 1)
end_date   = datetime(2024, 12, 31)

date_range = pd.date_range(start=start_date, end=end_date, freq='D')

dim_date = pd.DataFrame({
    'date_key'   : range(1, len(date_range) + 1),
    'date'       : date_range,
    'year'       : date_range.year,
    'quarter'    : date_range.quarter,
    'month'      : date_range.month,
    'month_name' : date_range.strftime('%B'),
    'week'       : date_range.isocalendar().week.astype(int),
    'day'        : date_range.day,
    'day_of_week': date_range.dayofweek,
    'day_name'   : date_range.strftime('%A'),
    'is_weekend' : date_range.dayofweek >= 5
})

dim_date['date'] = dim_date['date'].dt.strftime('%Y-%m-%d')

dim_date.to_csv('data/dim_date.csv', index=False)
print('dim_date:', dim_date.shape)
dim_date.head(3)
```

## Fact Tables

```python
# helper – map a date to its date_key
date_key_map = dict(zip(pd.to_datetime(dim_date['date']), dim_date['date_key']))

def random_date_keys(n, start='2019-01-01', end='2024-12-31'):
    """Return n random date_keys uniformly between start and end."""
    dates = pd.to_datetime(
        np.random.randint(
            pd.Timestamp(start).value,
            pd.Timestamp(end).value,
            size=n,
            dtype=np.int64
        )
    ).normalize()
    return dates.map(date_key_map).values
```

```python
# ── fact_sales ─────────────────────────────────────────────────────────────────
n_sales = 100_000

product_ids  = np.random.choice(dim_product['product_id'].values,  size=n_sales)
customer_ids = np.random.choice(dim_customer['customer_id'].values, size=n_sales)
date_keys    = random_date_keys(n_sales)

price_map = dict(zip(dim_product['product_id'], dim_product['unit_price']))
cost_map  = dict(zip(dim_product['product_id'], dim_product['unit_cost']))

quantity_sold = np.random.randint(1, 51, size=n_sales)
unit_prices   = np.array([price_map[p] for p in product_ids])
unit_costs_s  = np.array([cost_map[p]  for p in product_ids])
discount_pct  = np.round(np.random.choice([0, 0, 0, 5, 10, 15, 20], size=n_sales).astype(float), 2)

gross_revenue  = quantity_sold * unit_prices
discount_amt   = np.round(gross_revenue * discount_pct / 100, 2)
net_revenue    = np.round(gross_revenue - discount_amt, 2)
total_cost_s   = quantity_sold * unit_costs_s
profit         = np.round(net_revenue - total_cost_s, 2)
profit_margin  = np.round(np.where(net_revenue > 0, profit / net_revenue * 100, 0), 2)

fact_sales = pd.DataFrame({
    'sales_id'         : range(1, n_sales + 1),
    'date_key'         : date_keys,
    'product_id'       : product_ids,
    'customer_id'      : customer_ids,
    'quantity_sold'    : quantity_sold,
    'unit_price'       : unit_prices,
    'discount_pct'     : discount_pct,
    'order_number'     : [f'ORD-{str(i).zfill(7)}' for i in range(1, n_sales + 1)],
    'discount_amount'  : discount_amt,
    'gross_revenue'    : gross_revenue,
    'net_revenue'      : net_revenue,
    'total_cost'       : total_cost_s,
    'profit'           : profit,
    'profit_margin_pct': profit_margin
})

fact_sales.to_csv('data/fact_sales.csv', index=False)
print('fact_sales:', fact_sales.shape)
fact_sales.head(3)
```

```python
# ── fact_inventory ─────────────────────────────────────────────────────────────
n_inv = 8_000

fact_inventory = pd.DataFrame({
    'inventory_id'      : range(1, n_inv + 1),
    'date_key'          : random_date_keys(n_inv),
    'product_id'        : np.random.choice(dim_product['product_id'].values,  size=n_inv),
    'facility_id'       : np.random.choice(dim_facility['facility_id'].values, size=n_inv),
    'stock_level'       : np.random.randint(0,    2001, size=n_inv),
    'safety_stock_level': np.random.randint(50,   501,  size=n_inv),
    'reorder_point'     : np.random.randint(100,  601,  size=n_inv)
})

fact_inventory.to_csv('data/fact_inventory.csv', index=False)
print('fact_inventory:', fact_inventory.shape)
fact_inventory.head(3)
```

```python
# ── fact_procurement ───────────────────────────────────────────────────────────
n_proc = 3_000

order_date_keys = random_date_keys(n_proc, end='2024-11-30')
lead_times      = np.random.randint(5, 61, size=n_proc)

# compute delivery date_keys from order date + lead time
inv_date_key_map = {v: k for k, v in date_key_map.items()}
delivery_date_keys = []
for dk, lt in zip(order_date_keys, lead_times):
    delivery_dt = inv_date_key_map[dk] + timedelta(days=int(lt))
    delivery_dt = min(delivery_dt, datetime(2024, 12, 31))
    delivery_date_keys.append(date_key_map[delivery_dt])

proc_product_ids  = np.random.choice(dim_product['product_id'].values,   size=n_proc)
proc_supplier_ids = np.random.choice(dim_supplier['supplier_id'].values,  size=n_proc)
order_qty         = np.random.randint(50, 1001, size=n_proc)
proc_unit_costs   = np.array([cost_map[p] * np.random.uniform(0.85, 1.05) for p in proc_product_ids]).round(2)
total_cost_p      = np.round(order_qty * proc_unit_costs, 2)

fact_procurement = pd.DataFrame({
    'procurement_id'  : range(1, n_proc + 1),
    'order_date_key'  : order_date_keys,
    'product_id'      : proc_product_ids,
    'supplier_id'     : proc_supplier_ids,
    'order_quantity'  : order_qty,
    'unit_cost'       : proc_unit_costs,
    'lead_time_days'  : lead_times,
    'delivery_date_key': delivery_date_keys,
    'quality_score'   : np.round(np.random.uniform(3.0, 5.0, size=n_proc), 2),
    'po_number'       : [f'PO-{str(i).zfill(6)}' for i in range(1, n_proc + 1)],
    'total_cost'      : total_cost_p
})

fact_procurement.to_csv('data/fact_procurement.csv', index=False)
print('fact_procurement:', fact_procurement.shape)
fact_procurement.head(3)
```

```python
# ── fact_production ────────────────────────────────────────────────────────────
n_prod = 9_000

qty_produced   = np.random.randint(100, 1001, size=n_prod)
defect_rate    = np.round(np.random.uniform(0.0, 0.05, size=n_prod), 4)
defective_units = np.round(qty_produced * defect_rate).astype(int)

fact_production = pd.DataFrame({
    'production_id'   : range(1, n_prod + 1),
    'date_key'        : random_date_keys(n_prod),
    'product_id'      : np.random.choice(dim_product['product_id'].values,  size=n_prod),
    'facility_id'     : np.random.choice(dim_facility['facility_id'].values, size=n_prod),
    'quantity_produced': qty_produced,
    'defective_units' : defective_units,
    'batch_number'    : [f'BATCH-{str(i).zfill(6)}' for i in range(1, n_prod + 1)],
    'defect_rate_pct' : np.round(defect_rate * 100, 2)
})

fact_production.to_csv('data/fact_production.csv', index=False)
print('fact_production:', fact_production.shape)
fact_production.head(3)
```

```python
# ── fact_shipment ──────────────────────────────────────────────────────────────
n_ship = 12_000

carriers      = ['FedEx', 'UPS', 'DHL', 'USPS', 'Amazon Logistics']
statuses      = ['Delivered', 'Delivered', 'Delivered', 'In Transit', 'Delayed', 'Returned']
delay_reasons = ['Weather', 'Customs', 'Carrier Delay', 'Address Issue', None, None, None]

ship_date_keys = random_date_keys(n_ship, end='2024-11-30')
transit_days   = np.random.randint(1, 15, size=n_ship)

delivery_date_keys_s = []
for dk, td in zip(ship_date_keys, transit_days):
    delivery_dt = inv_date_key_map[dk] + timedelta(days=int(td))
    delivery_dt = min(delivery_dt, datetime(2024, 12, 31))
    delivery_date_keys_s.append(date_key_map[delivery_dt])

ship_qty     = np.random.randint(1, 201, size=n_ship)
ship_prod_ids = np.random.choice(dim_product['product_id'].values, size=n_ship)
weight_map   = dict(zip(dim_product['product_id'], dim_product['weight_kg']))
total_weight = np.round(ship_qty * np.array([weight_map[p] for p in ship_prod_ids]), 2)

ship_statuses     = np.random.choice(statuses, size=n_ship)
ship_delay_reasons = [
    random.choice(delay_reasons[:-2]) if s == 'Delayed' else None
    for s in ship_statuses
]

fact_shipment = pd.DataFrame({
    'shipment_id'      : range(1, n_ship + 1),
    'ship_date_key'    : ship_date_keys,
    'delivery_date_key': delivery_date_keys_s,
    'product_id'       : ship_prod_ids,
    'facility_id'      : np.random.choice(dim_facility['facility_id'].values, size=n_ship),
    'customer_id'      : np.random.choice(dim_customer['customer_id'].values, size=n_ship),
    'quantity'         : ship_qty,
    'carrier'          : np.random.choice(carriers, size=n_ship),
    'status'           : ship_statuses,
    'shipping_cost'    : np.round(np.random.uniform(5, 500, size=n_ship), 2),
    'total_weight_kg'  : total_weight,
    'tracking_number'  : [f'TRK-{str(i).zfill(8)}' for i in range(1, n_ship + 1)],
    'delay_reason'     : ship_delay_reasons
})

fact_shipment.to_csv('data/fact_shipment.csv', index=False)
print('fact_shipment:', fact_shipment.shape)
fact_shipment.head(3)
```
